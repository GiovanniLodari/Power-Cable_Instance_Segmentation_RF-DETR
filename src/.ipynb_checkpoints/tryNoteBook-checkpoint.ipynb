# Mask2Former Training with PolarHead - TTPLA Dataset

**Instance Segmentation + Polar Coordinates Prediction**

Questo notebook estende il training di Mask2Former introducendo una **PolarHead** per predire l'orientamento (rho, theta) dei cavi, oltre alla segmentazione.

### Novità in questo notebook:
- ✅ **PolarHead**: Nuova head neurale per regressione coordinate polari
- ✅ **Custom Architecture**: Estensione di `Mask2Former`, `MaskFormerHead` e `TransformerDecoder`
- ✅ **Polar Loss**: Integrazione di MSE Loss per le coordinate polari nel criterio di training
- ✅ **Validation Fix**: Risolto il bug di device mismatch durante l'inference

## 2. Dataset Loading & Mapper
Caricamento dataset e custom mapper (invariato).

```python
def load_ttpla_dataset(root_dir: str, split: str = 'train', val_split: float = 0.2):
    root_path = Path(root_dir)
    json_file = root_path / 'train.json'
    with open(json_file, 'r') as f:
        coco_data = json.load(f)
    
    all_images = sorted(coco_data['images'], key=lambda x: x['id'])
    num_val = int(len(all_images) * val_split)
    images = all_images[:num_val] if split == 'val' else all_images[num_val:]
    
    valid_img_ids = {img['id'] for img in images}
    img_to_anns = {}
    for ann in coco_data['annotations']:
        if ann['image_id'] in valid_img_ids:
            img_to_anns.setdefault(ann['image_id'], []).append(ann)
    
    dataset_dicts = []
    for img_info in images:
        record = {
            "file_name": str(root_path / img_info['file_name']),
            "image_id": img_info['id'],
            "height": img_info['height'],
            "width": img_info['width']
        }
        objs = []
        for ann in img_to_anns.get(img_info['id'], []):
            segmentation = ann['segmentation']
            if isinstance(segmentation, dict): # RLE
                from pycocotools import mask as mask_utils
                mask = mask_utils.decode(segmentation)
                contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                polygons = [c.flatten().tolist() for c in contours if len(c) >= 3]
            else:
                polygons = segmentation
            
            polar_coords = [0.0, 0.0]
            if 'polar_coordinates' in ann and len(ann['polar_coordinates']) > 0:
                p = ann['polar_coordinates'][0]
                polar_coords = [float(p.get('rho', 0.0)), float(p.get('theta', 0.0))]

            objs.append({
                "bbox": ann['bbox'],
                "bbox_mode": 1, # XYWH_ABS
                "segmentation": polygons,
                "category_id": 0,
                "polar_coordinates": polar_coords
            })
        record["annotations"] = objs
        dataset_dicts.append(record)
    return dataset_dicts

def custom_mapper(dataset_dict):
    dataset_dict = copy.deepcopy(dataset_dict)
    image = utils.read_image(dataset_dict["file_name"], format="BGR")
    
    augs = T.AugmentationList([
        T.RandomFlip(prob=0.5),
        T.ResizeShortestEdge(short_edge_length=(640, 672, 704, 736, 768, 800), max_size=1333, sample_style="choice"),
    ])
    aug_input = T.AugInput(image)
    transforms = augs(aug_input)
    image = aug_input.image
    dataset_dict["image"] = torch.as_tensor(image.transpose(2, 0, 1).astype("float32"))

    annos = [
        utils.transform_instance_annotations(obj, transforms, image.shape[:2])
        for obj in dataset_dict.pop("annotations")
    ]
    instances = utils.annotations_to_instances(annos, image.shape[:2])
    
    if len(annos) > 0:
        instances.gt_polars = torch.tensor([ann.get('polar_coordinates', [0.0, 0.0]) for ann in annos], dtype=torch.float32)
        
        # Convert PolygonMasks to BitMasks for Mask2Former compatibility
        if hasattr(instances, 'gt_masks') and isinstance(instances.gt_masks, PolygonMasks):
            instances.gt_masks = instances.gt_masks.get_bit_masks(instances.image_size[0], instances.image_size[1])
            instances.gt_masks = BitMasks(instances.gt_masks)

    instances = utils.filter_empty_instances(instances)
    
    # Convert BitMasks to tensor
    if hasattr(instances, 'gt_masks') and isinstance(instances.gt_masks, BitMasks):
        instances.gt_masks = instances.gt_masks.tensor

    dataset_dict["instances"] = instances
    return dataset_dict
```

## 3. Architecture Definition
Qui definiamo le classi custom per integrare la PolarHead:
1. `PolarHead`: Modulo MLP per predire (rho, theta)
2. `PolarTransformerDecoder`: Estende il decoder per ritornare gli embeddings
3. `PolarMaskFormerHead`: Usa il nuovo decoder e la PolarHead
4. `PolarSetCriterion`: Calcola anche la loss polare
5. `PolarMaskFormer`: Meta-architettura che assembla tutto

```python
# 3.1 PolarHead
class PolarHead(nn.Module):
    def __init__(self, input_dim, hidden_dim=256):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, 2)  # rho, theta
        )
    
    def forward(self, x):
        return self.layers(x)

# 3.2 PolarTransformerDecoder
class PolarTransformerDecoder(MultiScaleMaskedTransformerDecoder):
    def forward(self, x, mask_features, mask=None):
        # Copia quasi esatta del forward originale, ma ritorna anche 'pred_polars'
        assert len(x) == self.num_feature_levels
        src = []
        pos = []
        size_list = []
        del mask

        for i in range(self.num_feature_levels):
            size_list.append(x[i].shape[-2:])
            pos.append(self.pe_layer(x[i], None).flatten(2))
            src.append(self.input_proj[i](x[i]).flatten(2) + self.level_embed.weight[i][None, :, None])
            pos[-1] = pos[-1].permute(2, 0, 1)
            src[-1] = src[-1].permute(2, 0, 1)

        _, bs, _ = src[0].shape
        query_embed = self.query_embed.weight.unsqueeze(1).repeat(1, bs, 1)
        output = self.query_feat.weight.unsqueeze(1).repeat(1, bs, 1)

        predictions_class = []
        predictions_mask = []
        predictions_polar = [] # NEW

        # Initial prediction
        outputs_class, outputs_mask, attn_mask, decoder_output = self.forward_prediction_heads_with_embed(output, mask_features, size_list[0])
        predictions_class.append(outputs_class)
        predictions_mask.append(outputs_mask)
        predictions_polar.append(decoder_output) # We will apply PolarHead later or return embeddings

        for i in range(self.num_layers):
            level_index = i % self.num_feature_levels
            attn_mask[torch.where(attn_mask.sum(-1) == attn_mask.shape[-1])] = False
            output = self.transformer_cross_attention_layers[i](
                output, src[level_index],
                memory_mask=attn_mask,
                memory_key_padding_mask=None,
                pos=pos[level_index], query_pos=query_embed
            )
            output = self.transformer_self_attention_layers[i](
                output, tgt_mask=None,
                tgt_key_padding_mask=None,
                query_pos=query_embed
            )
            output = self.transformer_ffn_layers[i](output)

            outputs_class, outputs_mask, attn_mask, decoder_output = self.forward_prediction_heads_with_embed(output, mask_features, size_list[(i + 1) % self.num_feature_levels])
            predictions_class.append(outputs_class)
            predictions_mask.append(outputs_mask)
            predictions_polar.append(decoder_output)

        out = {
            'pred_logits': predictions_class[-1],
            'pred_masks': predictions_mask[-1],
            'pred_embeddings': predictions_polar[-1], # Return final embeddings
            'aux_outputs': self._set_aux_loss(predictions_class, predictions_mask)
        }
        return out

    def forward_prediction_heads_with_embed(self, output, mask_features, attn_mask_target_size):
        decoder_output = self.decoder_norm(output)
        decoder_output = decoder_output.transpose(0, 1)
        outputs_class = self.class_embed(decoder_output)
        mask_embed = self.mask_embed(decoder_output)
        outputs_mask = torch.einsum("bqc,bchw->bqhw", mask_embed, mask_features)

        attn_mask = F.interpolate(outputs_mask, size=attn_mask_target_size, mode="bilinear", align_corners=False)
        attn_mask = (attn_mask.sigmoid().flatten(2).unsqueeze(1).repeat(1, self.num_heads, 1, 1).flatten(0, 1) < 0.5).bool()
        attn_mask = attn_mask.detach()

        return outputs_class, outputs_mask, attn_mask, decoder_output

# 3.3 PolarMaskFormerHead
@SEM_SEG_HEADS_REGISTRY.register()
class PolarMaskFormerHead(MaskFormerHead):
    @configurable
    def __init__(self, input_shape, **kwargs):
        super().__init__(input_shape, **kwargs)
        # Add PolarHead
        # Hidden dim is typically 256 for Swin-Base config
        self.polar_head = PolarHead(256)

    @classmethod
    def from_config(cls, cfg, input_shape):
        ret = super().from_config(cfg, input_shape)
        # Override predictor with our custom one
        ret["transformer_predictor"] = PolarTransformerDecoder(
            cfg.MODEL.SEM_SEG_HEAD.CONVS_DIM if cfg.MODEL.MASK_FORMER.TRANSFORMER_IN_FEATURE == "multi_scale_pixel_decoder" else input_shape[cfg.MODEL.MASK_FORMER.TRANSFORMER_IN_FEATURE].channels,
            mask_classification=True,
            num_classes=cfg.MODEL.SEM_SEG_HEAD.NUM_CLASSES,
            hidden_dim=cfg.MODEL.MASK_FORMER.HIDDEN_DIM,
            num_queries=cfg.MODEL.MASK_FORMER.NUM_OBJECT_QUERIES,
            nheads=cfg.MODEL.MASK_FORMER.NHEADS,
            dim_feedforward=cfg.MODEL.MASK_FORMER.DIM_FEEDFORWARD,
            dec_layers=cfg.MODEL.MASK_FORMER.DEC_LAYERS - 1,
            pre_norm=cfg.MODEL.MASK_FORMER.PRE_NORM,
            mask_dim=cfg.MODEL.SEM_SEG_HEAD.MASK_DIM,
            enforce_input_project=cfg.MODEL.MASK_FORMER.ENFORCE_INPUT_PROJ,
        )
        return ret

    def forward(self, features, mask=None):
        outputs = self.layers(features, mask)
        # outputs has 'pred_embeddings'
        embeddings = outputs.pop('pred_embeddings')
        # Predict polars
        pred_polars = self.polar_head(embeddings)
        outputs['pred_polars'] = pred_polars
        return outputs

# 3.4 PolarSetCriterion
class PolarSetCriterion(SetCriterion):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.losses.append("polar") # Add polar loss to list

    def loss_polar(self, outputs, targets, indices, num_masks):
        assert "pred_polars" in outputs
        src_polars = outputs["pred_polars"]
        src_idx = self._get_src_permutation_idx(indices)
        
        src_polars = src_polars[src_idx]
        target_polars = torch.cat([t["polar_coordinates"][i] for t, (_, i) in zip(targets, indices)], dim=0)
        
        loss_polar = F.mse_loss(src_polars, target_polars, reduction='none')
        loss_polar = loss_polar.sum() / num_masks
        return {"loss_polar": loss_polar}

    def get_loss(self, loss, outputs, targets, indices, num_masks):
        if loss == 'polar':
            return self.loss_polar(outputs, targets, indices, num_masks)
        return super().get_loss(loss, outputs, targets, indices, num_masks)

# 3.5 PolarMaskFormer (Meta Arch)
from mask2former.maskformer_model import MaskFormer

@META_ARCH_REGISTRY.register()
class PolarMaskFormer(MaskFormer):
    @classmethod
    def from_config(cls, cfg):
        ret = super().from_config(cfg)
        # Replace criterion with PolarSetCriterion
        matcher = HungarianMatcher(
            cost_class=cfg.MODEL.MASK_FORMER.CLASS_WEIGHT,
            cost_mask=cfg.MODEL.MASK_FORMER.MASK_WEIGHT,
            cost_dice=cfg.MODEL.MASK_FORMER.DICE_WEIGHT,
            num_points=cfg.MODEL.MASK_FORMER.TRAIN_NUM_POINTS,
        )
        weight_dict = {"loss_ce": cfg.MODEL.MASK_FORMER.CLASS_WEIGHT, 
                       "loss_mask": cfg.MODEL.MASK_FORMER.MASK_WEIGHT, 
                       "loss_dice": cfg.MODEL.MASK_FORMER.DICE_WEIGHT,
                       "loss_polar": 1.0} # Polar Loss Weight
        
        if cfg.MODEL.MASK_FORMER.DEEP_SUPERVISION:
            dec_layers = cfg.MODEL.MASK_FORMER.DEC_LAYERS
            aux_weight_dict = {}
            for i in range(dec_layers - 1):
                aux_weight_dict.update({k + f"_{i}": v for k, v in weight_dict.items()})
            weight_dict.update(aux_weight_dict)

        criterion = PolarSetCriterion(
            ret["sem_seg_head"].num_classes,
            matcher=matcher,
            weight_dict=weight_dict,
            eos_coef=cfg.MODEL.MASK_FORMER.NO_OBJECT_WEIGHT,
            losses=["labels", "masks", "polar"],
            num_points=cfg.MODEL.MASK_FORMER.TRAIN_NUM_POINTS,
            oversample_ratio=cfg.MODEL.MASK_FORMER.OVERSAMPLE_RATIO,
            importance_sample_ratio=cfg.MODEL.MASK_FORMER.IMPORTANCE_SAMPLE_RATIO,
        )
        ret["criterion"] = criterion
        return ret

    def prepare_targets(self, targets, images):
        h_pad, w_pad = images.tensor.shape[-2:]
        new_targets = []
        for targets_per_image in targets:
            gt_masks = targets_per_image.gt_masks
            # Assicura device corretto
            padded_masks = torch.zeros((gt_masks.shape[0], h_pad, w_pad), dtype=gt_masks.dtype, device=gt_masks.device)
            padded_masks[:, : gt_masks.shape[1], : gt_masks.shape[2]] = gt_masks
            
            new_targets.append(
                {
                    "labels": targets_per_image.gt_classes,
                    "masks": padded_masks,
                    "polar_coordinates": targets_per_image.gt_polars.to(gt_masks.device) # <--- Explicit device move
                }
            )
        return new_targets
```

## 4. Custom Train & Visualize

```python
# --- CELLA 4: Custom Trainer e Visualization Hook ---
from detectron2.engine import HookBase
from detectron2.utils.events import get_event_storage
from rich.console import Console
from rich.table import Table
import matplotlib.pyplot as plt
from IPython.display import clear_output

class RichLossPlotterHook(HookBase):
    """
    Questo Hook viene chiamato alla fine di ogni iterazione (o step).
    Si occupa di:
    1. Leggere le loss correnti dallo storage di Detectron2.
    2. Aggiornare una tabella Rich.
    3. Disegnare i grafici live con Matplotlib.
    """
    def __init__(self, plot_interval=20):
        self._plot_interval = plot_interval
        # Dizionario per memorizzare la storia delle loss
        self._history = {"total_loss": [], "loss_polar": [], "loss_mask": [], "loss_ce": [], "loss_dice": []}
        self._console = Console()
        
    def after_step(self):
        storage = get_event_storage()
        
        # Eseguiamo l'aggiornamento solo ogni 'plot_interval' step per non rallentare troppo
        if storage.iter % self._plot_interval == 0:
            latest_metrics = storage.latest()
            
            # Salviamo i dati per i grafici
            if "total_loss" in latest_metrics:
                self._history["total_loss"].append(latest_metrics["total_loss"][0])
            if "loss_polar" in latest_metrics:
                self._history["loss_polar"].append(latest_metrics["loss_polar"][0])
            if "loss_mask" in latest_metrics:
                self._history["loss_mask"].append(latest_metrics["loss_mask"][0])
            
            # --- Visualizzazione ---
            clear_output(wait=True) # Pulisce l'output della cella precedente
            
            # 1. Creiamo la Tabella Rich
            table = Table(title=f"Training Status - Iter {storage.iter}")
            table.add_column("Metric", style="cyan")
            table.add_column("Value", style="magenta")
            
            for k, v in latest_metrics.items():
                if "loss" in k: # Mostriamo solo le metriche di loss
                    val = v[0]
                    table.add_row(k, f"{val:.4f}")
            
            self._console.print(table)
            
            # 2. Disegniamo i Grafici
            if len(self._history["total_loss"]) > 1:
                fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
                
                # Grafico 1: Loss Totale
                ax1.plot(self._history["total_loss"], label="Total Loss", color='red')
                ax1.set_title("Total Loss Trend")
                ax1.set_xlabel(f"Steps (x{self._plot_interval})")
                ax1.set_ylabel("Loss")
                ax1.grid(True, alpha=0.3)
                ax1.legend()
                
                # Grafico 2: Loss Specifiche (Polar vs Mask)
                if len(self._history["loss_polar"]) > 0:
                    ax2.plot(self._history["loss_polar"], label="Polar Loss", color='blue')
                if len(self._history["loss_mask"]) > 0:
                    ax2.plot(self._history["loss_mask"], label="Mask Loss", color='green')
                
                ax2.set_title("Component Losses")
                ax2.set_xlabel(f"Steps (x{self._plot_interval})")
                ax2.grid(True, alpha=0.3)
                ax2.legend()
                
                plt.show()

# Definiamo il Trainer che usa questo Hook
class PolarTrainer(DefaultTrainer):
    def build_hooks(self):
        # Carica gli hook standard di Detectron2 (checkpoint, timer, etc.)
        hooks = super().build_hooks()
        # Aggiunge il nostro hook personalizzato
        hooks.insert(-1, RichLossPlotterHook(plot_interval=20)) 
        return hooks

print("✅ Custom Trainer definito con successo.")
```

## 5. Configurazione & Training
Setup che usa la nuova architettura `PolarMaskFormer`.

```python
# 1. Registrazione Dataset (Assicurati che il path sia corretto)
DatasetCatalog.clear()
MetadataCatalog.clear()
data_root = "/mnt/c/Users/giova/Downloads/progetto_cv_2025_2026/train" # <--- CONTROLLA QUESTO PATH

for split in ['train', 'val']:
    DatasetCatalog.register(f"ttpla_{split}", lambda s=split: load_ttpla_dataset(data_root, s))
    MetadataCatalog.get(f"ttpla_{split}").set(thing_classes=["cable"])

# 2. Funzione di Configurazione
def setup_polar_config(output_dir):
    cfg = get_cfg()
    add_maskformer2_config(cfg)
    
    # Path del config base (Assicurati che il file esista nel path indicato)
    config_file = project_root / 'Mask2Former' / 'configs' / 'coco/instance-segmentation/swin/maskformer2_swin_base_IN21k_384_bs16_50ep.yaml'
    
    # --- FIX: Allow new keys for legacy configs ---
    cfg.set_new_allowed(True)
    cfg.merge_from_file(str(config_file))
    cfg.set_new_allowed(False)
    # ----------------------------------------------

    cfg.SOLVER.CLIP_GRADIENTS.CLIP_TYPE = "norm"
    cfg.SOLVER.CLIP_GRADIENTS.CLIP_VALUE = 0.01
    
    # Custom Architecture
    cfg.MODEL.META_ARCHITECTURE = "PolarMaskFormer"
    cfg.MODEL.SEM_SEG_HEAD.NAME = "PolarMaskFormerHead"
    
    # Dataset
    cfg.DATASETS.TRAIN = ("ttpla_train",)
    cfg.DATASETS.TEST = ("ttpla_val",)
    cfg.DATALOADER.NUM_WORKERS = 2 
    
    # Pesi iniziali (Backbone Swin pre-addestrato)
    cfg.MODEL.WEIGHTS = "/mnt/c/Users/giova/Desktop/VisionTranformer/tools/swin_base_converted.pkl"
    
    # Parametri Modello
    cfg.MODEL.MASK_FORMER.NUM_OBJECT_QUERIES = 100
    cfg.MODEL.SEM_SEG_HEAD.NUM_CLASSES = 1 # Solo cavi
    
    # Parametri Training
    cfg.SOLVER.IMS_PER_BATCH = 2      # Abbassa a 1 se hai errori di memoria (OOM)
    cfg.SOLVER.BASE_LR = 0.0001
    cfg.SOLVER.MAX_ITER = 1000        # Numero totale di step
    cfg.SOLVER.CHECKPOINT_PERIOD = 200 # Salva ogni 200 step
    
    cfg.OUTPUT_DIR = output_dir
    os.makedirs(cfg.OUTPUT_DIR, exist_ok=True)
    
    return cfg

# 3. Impostazioni di Esecuzione
OUTPUT_DIR = "./output_polar_mask2former"
RESUME = True  # <--- METTI 'True' PER RIPRENDERE IL TRAINING, 'False' PER RICOMINCIARE

# Setup
cfg = setup_polar_config(OUTPUT_DIR)

# Creazione del Trainer (Usa la nostra classe PolarTrainer definita nella Cella 4)
trainer = PolarTrainer(cfg)

# Avvio
print(f"🚀 Avvio Training (Resume={RESUME})...")
print("I grafici appariranno qui sotto ogni 20 step.")

# Se resume=True, cerca automaticamente 'last_checkpoint' nella cartella output
trainer.resume_or_load(resume=RESUME)
trainer.train()
```

## 6. Validation & Inference
Visualizzazione risultati con fix per il device mismatch.

```python
def visualize_polar_predictions(predictor, image_path, output_path, score_threshold=0.5):
    img = cv2.imread(image_path)
    outputs = predictor(img)
    
    # FIX: Move instances to CPU before filtering
    instances = outputs["instances"].to("cpu")
    instances = instances[instances.scores > score_threshold]
    
    # Print predicted polar coords
    if hasattr(instances, 'pred_polars'):
        print("Predicted Polars (rho, theta):", instances.pred_polars)
    
    v = Visualizer(
        img[:, :, ::-1],
        metadata=MetadataCatalog.get("ttpla_val"),
        scale=1.0,
        instance_mode=ColorMode.IMAGE_BW
    )
    
    vis = v.draw_instance_predictions(instances)
    vis_img = vis.get_image()[:, :, ::-1]
    cv2.imwrite(output_path, vis_img)
    return vis_img

# Load final model
cfg.MODEL.WEIGHTS = os.path.join(cfg.OUTPUT_DIR, "model_final.pth")
predictor = DefaultPredictor(cfg)

val_dataset = DatasetCatalog.get("ttpla_val")
if len(val_dataset) > 0:
    img_path = val_dataset[0]['file_name']
    print(f"Visualizzazione su: {img_path}")
    vis_img = visualize_polar_predictions(predictor, img_path, "polar_prediction.jpg")
    
    import matplotlib.pyplot as plt
    plt.figure(figsize=(12, 8))
    plt.imshow(vis_img[:, :, ::-1])
    plt.axis('off')
    plt.show()
```
